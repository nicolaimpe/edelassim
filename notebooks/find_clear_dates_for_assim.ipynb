```python
import xarray as xr
from edelassim.observations import valid_snow_cover_fraction_viirs_mf
import pandas as pd

vi = xr.open_dataset("/home/imperatoren/work/edelweiss_assimilation/observations/grandesrousses250m/meteofrance/spatial.nc")
va_vi = valid_snow_cover_fraction_viirs_mf(vi.data_vars["snow_cover_fraction"].sel(time=slice("2021-11", "2022-07")))
forest_mask_path = "/home/imperatoren/work/edelweiss_assimilation/data/grandesrousses250m/auxiliary/forest_mask/forest_mask_corine_grandesrousses_max.nc"
glacier_mask_path = "/home/imperatoren/work/edelweiss_assimilation/data/grandesrousses250m/auxiliary/glacier_mask/glacier_mask_glims_2022_grandesrousses.nc"
glacier_mask = xr.open_dataset(glacier_mask_path).data_vars["__xarray_dataarray_variable__"]
forest_mask = xr.open_dataset(forest_mask_path).sel(band=1).data_vars["__xarray_dataarray_variable__"]
mask = glacier_mask + forest_mask
va_vi_masked = va_vi.where(1 - mask)
assim_dates = va_vi_masked.where((va_vi_masked.count(dim=("x", "y")) / (1 - mask).sum() > 0.7), drop=True).time
# [print(f"'{a}'", end=",") for a in [pd.Timestamp(ad).strftime("%Y-%m-%d") for ad in assim_dates.time.values]]
[print(f"{a}", end=" ") for a in [pd.Timestamp(ad).strftime("%Y%m%d12") for ad in assim_dates.time.values]]
```
